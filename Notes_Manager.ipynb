```python

notes = []

def add():
    notes.append(input("Enter note: "))
    print("Note added!")

def view():
    print(notes if notes else "No notes found!")

def delete():
    n = input("Enter note to delete: ")
    if n in notes:
        notes.remove(n)
        print("Deleted!")
    else:
        print("Not found!")

def main():
    while True:
        print("\n1.Add  2.View  3.Delete  4.Exit")
        ch = input("Choice: ")

        if ch == "1": add()
        elif ch == "2": view()
        elif ch == "3": delete()
        elif ch == "4": break
        else: print("Invalid choice!")

main()
```
